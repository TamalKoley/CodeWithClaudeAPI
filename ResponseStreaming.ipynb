### How to Stream response

```python
from anthropic import Anthropic;

def chatStream(message):
    client=Anthropic();
    model='claude-sonnet-4-5'
    with client.messages.stream(
        model=model,
        max_tokens=1000,
        messages=[{
            "role":"user",
            "content": message
        }],
    ) as stream:
        for text in stream.text_stream:
            print(text,end="");

chatStream("give a 1 sentence fake database desciption");



```
