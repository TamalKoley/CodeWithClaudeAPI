### A Simple Chatbot

```python
from anthropic import Anthropic;

def addMessage(messageList,message,role):
    msg={"role":role,"content":message};
    messageList.append(msg);
    #print(messageList);

def anthropicChat(client,model,messageList):
    answer=client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messageList
    );
    return answer.content[0].text;

def anthropicChatBot():
    anthropiClient=Anthropic();
    model="claude-sonnet-4-5";
    messageList=[];

    while True:
        userQuery=input();
        if userQuery.lower()=="stop":
             break;
        else:
            addMessage(messageList,userQuery,"user");
            answer=anthropicChat(anthropiClient,model,messageList);
            addMessage(messageList,answer,"assistant");
            print(userQuery);
            print(answer);
        
```

```python
anthropicChatBot()
```
