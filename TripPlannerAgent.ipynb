### A Helpful Trip Planner

````python
from dotenv import load_dotenv;
from anthropic import Anthropic;
import json;
````

````python
def chat(messageList,stopSequence=None,model='claude-haiku-4-5'):
    stop_Sequence=[];
    if stopSequence:
        stop_Sequence.append(stopSequence)
    client=Anthropic();
    response=client.messages.create(
        model=model,
        max_tokens=20000,
        messages=messageList,
        stop_sequences=stop_Sequence
    )
    return response.content[0].text
````

````python
def addMessage(messageList,message,role):
    msg={
        "role": role,
        "content": message
    }
    messageList.append(msg);
    print(messageList)
    return messageList;
````

````python
def createTripPlan(trip_description):
    assistantPrompt="""
    You are a exellecnt Trip Planner. Your task as follows:
    1. Analyze and understand the trip requirement
    2. create a trip plan according to the requirement.
    3. trip plan must contain all the main attraction places along with other attractions.
    4. trip plan must contain breakfast and lunch breaks along with best places to eat on the way.
    5. trip plan must have all the start time and end time for each step.
    6. trip plan must have weather details and necessary clothing requirement for it.
    7. trip plan must have parking charges, entry fees if applicable and its amount and the place to buy the same.


    Output:
    Output should be in json format  example :
    ```json
    {
        "Day 1" :[{
            "weather" : "weather details"
            "activity" : "start from home and reach to ...",
            "timing " : "10:00 AM - 11:30 AM"
            "activity_details " : "explore ...."
        },
        {
            "weather" : "weather details"
            "activity" : "start from home and reach to ...",
            "timing " : "10:00 AM - 11:30 AM"
            "activity_details " : "explore ...."
        }
          ...additional
        ],
        "Day 2"[
          ... additional
        ]
    }
    ```
    
    * Output should be concise and accurate
    * Output should be short and crisp but also need to have all necessary details
    * Output must no be too long with informating overflowing.
    """
    messageList=[];
    addMessage(messageList,trip_description,"user");
    addMessage(messageList,assistantPrompt.strip(),"assistant");
    addMessage(messageList,"```json","assistant");
    answer=chat(messageList,"```")
    return answer


    
````

````python
myTripPrompt="""
person : 4 adults and 1 child
Time : 8th Oct 26 - 10th oct 26
trip : 3 days tour from Halifax to cape Breton via car , frist day stay at chetticamp and second day stay at near juniper mountain sydney and must do skyline trail roaming on seconday day first slot
also must include stop at famous seafood chowder places on the way once a day
"""

plan=createTripPlan(myTripPrompt)
````
