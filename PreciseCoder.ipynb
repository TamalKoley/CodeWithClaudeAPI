### A precise coder using system prompts

```python
from anthropic import Anthropic;

def chat(message,systemPrompt=None):
    client=Anthropic();
    model="claude-sonnet-4-5";

    params={
        "model":model,
        "max_tokens":1000,
        "messages": [{
            "role":"user",
            "content":message
        }]
    }

    if systemPrompt:
        params["system"]=systemPrompt;
    
    answer=client.messages.create(
        **params
    );
    return answer.content[0].text;



```

```python
message=[];
answer=chat("write a code in python to check if a string contains duplicate letter","You are a python engineer and you are a precise coder")
print(answer);
```
