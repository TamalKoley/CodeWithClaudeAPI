### Playing with tempareture
#### Low tempareture --> deterministic
#### High tempareture --> creative

```python
from anthropic import Anthropic;

def chat(message):
    client=Anthropic();
    model='claude-sonnet-4-5';
    params={
        "model":model,
        "max_tokens":1000,
        "messages":[{
            "role":"user",
            "content":message
        }]
    }

    answer=client.messages.create(**params);
    return answer.content[0].text

idea1=chat("give me a new movie idea in one line")
idea2=chat("give me a new movie idea in one line")
idea3=chat("give me a new movie idea in one line")
print(idea1)
print(idea2)
print(idea3)

```

```python
from anthropic import Anthropic;

def chat(message):
    client=Anthropic();
    model='claude-sonnet-4-5';
    params={
        "model":model,
        "max_tokens":1050,
        "messages":[{
            "role":"user",
            "content":message
        }],
        "thinking":{
            "type":"enabled",
            "budget_tokens" : 1024
        }
    }

    answer=client.messages.create(**params);
    return answer

idea1=chat("give me a new movie idea in one line")
idea2=chat("give me a new movie idea in one line")
idea3=chat("give me a new movie idea in one line")
print(idea1)
print(idea2)
print(idea3)
```
