### to get a structured Output

````python
from anthropic import Anthropic;

def chat(message):
    client=Anthropic();
    model='claude-sonnet-4-5';
    answer=client.messages.create(
        model=model,
        max_tokens=1000,
        messages=[{
            "role":"user",
            "content":message
        }]
    )
    return answer.content[0].text

answer=chat("write a very short aws connection bridge in json")
print(answer)

````

````python
### With formatting

from anthropic import Anthropic;

def chat(message):
    client=Anthropic();
    model='claude-sonnet-4-5';
    answer=client.messages.create(
        model=model,
        max_tokens=1000,
        messages=[{
            "role":"user",
            "content":message
        },{
            "role": "assistant",
            "content" : "```json" ##messgae prefilling
        }],
        stop_sequences=["```"]
    )
    return answer.content[0].text

answer=chat("write a very short aws connection bridge in json")
print(answer)
````

````python
### With formatting

from anthropic import Anthropic;

def chat(message):
    client=Anthropic();
    model='claude-sonnet-4-5';
    answer=client.messages.create(
        model=model,
        max_tokens=1000,
        messages=[{
            "role":"user",
            "content":message
        },{
            "role": "assistant",
            "content" : "```bash"  #bash means only bash command will claude pick
        }],
        stop_sequences=["```"]
    )
    return answer.content[0].text

answer=chat("write 3 different very short aws commands")
print(answer)
````
