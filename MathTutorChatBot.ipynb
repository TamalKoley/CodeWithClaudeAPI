### A Math Tutor Assistant with System prompt customization

```python
from anthropic import Anthropic;

def addMessage(messageList,message,role):
    msg={"role":role,"content":message};
    messageList.append(msg);

def chat(messageList,model,client,systemPrompt=None):
    params={
        "model":model,
        "max_tokens":1000,
        "messages":messageList
    }

    if systemPrompt:
        params["system"]=systemPrompt;

    answer=client.messages.create(**params);
    return answer.content[0].text;



def mathTutor():
    anthropicClient=Anthropic();
    model="claude-sonnet-4-5";
    messageList=[];
    systemPrompt="""
            You are a patient math Tutor. You don't give the reslution directly to the user. Instead you help the 
            users with hints to solve the problem step by step. so that user can have the oppertunity to  think and solve it . If user is not 
            able to solve it at all after all the guidance then only you solve it for the user.
            """
    while True:
        userQuery=input(">>");
        if userQuery.lower()=="stop":
            break;
        else:
            addMessage(messageList,userQuery,"user");
            answer=chat(messageList,model,anthropicClient,systemPrompt);
            addMessage(messageList,answer,"assistant");
            print(userQuery);
            print("--------------------");
            print(answer);
```

```python
mathTutor()
```
